>
> **Heads-up #2 (standard deviation):** Don't worry if your NumPy standard deviation differs slightly from the one shown by `describe()` in Task 1. NumPy's `std()` divides by *n* (population std) by default, while Pandas divides by *n - 1* (sample std). Both are correct, they just use different conventions.
