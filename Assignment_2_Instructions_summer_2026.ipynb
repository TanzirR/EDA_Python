````python
import csv

# Part 2: YOUR CODE HERE
def load_data(filepath:str) -> list:
    with open(filepath, "r") as file:
        contents = csv.reader(file, delimiter=",")
        contents = list(contents)
        header = contents[0]
        rows = contents[1:]
        return header, rows   


header, rows = load_data("./data/goodreads_books.csv")
#print header
print(f"header: {header}")
#print first 5 rows
for i in range(len(rows)):
    if i == 5:
        break
    else:
        print(rows[i])

#print total number of rows loaded
print(f"Total number of rows loaded: {len(rows)}")
````

## Part 3: Extract and Clean a Column (20 points)

Write a function `extract_and_clean(rows, col_index, target_type)` that:
1. Takes a list of rows, a column index (integer), and a target type (`'int'`, `'float'`, or `'str'`).
2. For each row:
   - Extracts the value at `col_index` (if the row has enough columns; otherwise skip it).
   - Strips whitespace from the value.
   - Attempts to convert to the target type.
   - If conversion fails or the value is empty, appends `None` instead.
3. Returns a **list** of cleaned/converted values and a **count of how many values were invalid**.

**Hint:** Use `try` / `except` to catch conversion errors. For example:
```python
try:
    result = int(value)
except ValueError:
    result = None
```

**Function signature:**
```python
def extract_and_clean(rows, col_index, target_type):
    # Your code here
    return cleaned_data, invalid_count
```

**After defining the function:**
- Use the function to extract and clean 2-3 columns from your data. Look at your header to decide which column index to use for each field.  
  For example, if your header is `['id', 'year', 'category', 'value']`, then `year` is at index `1`, `category` at index `2`, etc.
- Print the number of invalid values for each column.
- Print the first 10 cleaned values from each column.

````python
# Part 3: YOUR CODE HERE

def extract_and_clean(rows: list, col_index: int, target_type):
    cleaned_data = []
    invalid_count = 0
   
    for row in rows:
        #if index out of bound continue to next row:
        if len(row) - 1 < col_index:
            continue
        #fix white spaces
        stripped_value = row[col_index].strip()
        #if the col is empty, append None and continue to next row: 
        if stripped_value == "":
            cleaned_data.append(None)
            invalid_count +=1
            continue
        #type conversion:
        if target_type == "int":
            try:
                cleaned_data.append(int(stripped_value))
            except ValueError:
                cleaned_data.append(None)
                invalid_count += 1
        elif target_type == "float":
            try:
                cleaned_data.append(float(stripped_value))
            except ValueError:
                cleaned_data.append(None)
                invalid_count += 1
        elif target_type == "str":
            cleaned_data.append(str(stripped_value))
    return cleaned_data, invalid_count

#print out num of invalid values for col such as title(str), average rating (float), ratings_count(int)
title_cleaned_data, title_invalid_count = extract_and_clean(rows,header.index("title"),"str")
avg_rating_cleaned_data, avg_rating_invalid_count = extract_and_clean(rows, header.index("average_rating"), "float")
ratings_count_cleaned_data, ratings_count_invalid_count = extract_and_clean(rows, header.index("ratings_count"), "int")
print(f'Invalid counts in the title column: {ratings_count_invalid_count}')
print(f'Invalid counts in the average rating column: {avg_rating_invalid_count}')
print(f'Invalid counts in the ratings count column: {ratings_count_invalid_count}')
#print first 10 clean values from each col
print(f"First 10 cleaned values from title column: {title_cleaned_data[:10]}")
print(f"First 10 cleaned values from average ratings column: {title_cleaned_data[:10]}")
print(f"First 10 cleaned values from ratings count column: {title_cleaned_data[:10]}")

````
