## Part 5: File Operations and Dataset Loading (20 points)

### Task 5.1: Create Dataset Info File (10 points)
Create a text file called `dataset_info.txt` that contains:
- Dataset name
- Number of rows
- Number of columns
- Data source
- Date you started working with it (you can use today's date as a string)

Use the `with open()` statement with `mode='w'` to write the file.

```python
# Task 5.1: YOUR CODE HERE
with open("dataset_info.txt", mode="w") as file:
    file.write(
        f"Dataset name: {dataset_name}\n"
        f"Number of rows: {rows}\n"
        f"Number of columns: {columns}\n"
        f"Data source: {dataset_source}\n"
        f"Date: 25 June 2026"
    )




```

### Task 5.2: Read and Print the File (10 points)
Read the `dataset_info.txt` file you just created and print its contents.

```python
# Task 5.2: YOUR CODE HERE
with open("dataset_info.txt", mode = "r") as file:
    contents = file.read()
print(contents)



```
